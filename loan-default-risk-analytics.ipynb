```python
# Risk Score Logic: Rule-Based High Risk Classification
```

```python
df[['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3', 'TARGET']].corr()
```

```python
def risk_score(row):
    if row['EXT_SOURCE_3'] < 0.3 and row['AGE_YEARS'] < 30 and row['REGION_RATING_CLIENT'] == 3:
                return 'High Risk'
    else:
        return 'Low Risk'
        
df['RISK_CATEGORY'] = df.apply(risk_score, axis=1)
```

```python
df['RISK_CATEGORY'].value_counts()
```

```python
df.groupby('RISK_CATEGORY')['TARGET'].mean() * 100
```
